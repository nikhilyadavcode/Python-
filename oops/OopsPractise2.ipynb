```python
class Employee:
    def __init__(self,name,salary,department):
        self.name=name
        self.salary=salary
        self.department=department

    def display_info(self):
        print(f"Name:{self.name} Salary:{self.salary} department {self.department}")

    def annual_bonus(self):
        bonus=self.salary*0.10
        return bonus
    
    def promote(self,new_department,increment):
        self.department=new_department
        self.salary+=increment
        print(f"{self.salary} has been promoted to {self.department} with new salary {self.salary}")

    emp1=Employee("Mohit",70000,"CSE")
    emp1.display_info()
    print("Annual Bonus is:", emp1.annual_bonus())
    emp1.promote("Senior IT",10000)
    emp1.display_info()
```

```python
class Student:
    def __init__(self, name, salary, experience, skills):
        self.name = name
        self.base_salary = salary         # Salary attribute ka naam change kiya
        self.experience = experience
        self.skills = skills

    def calculate_salary(self):
        salary = self.base_salary

        # Experience ke basis par increment
        if self.experience >= 5:
            salary += 20000
        elif self.experience >= 2:
            salary += 10000

        # Skills ke basis par increment
        if "Python" in self.skills:
            salary += 5000
        if "Java" in self.skills:
            salary += 4000
        if "Data Science" in self.skills:
            salary += 15000

        return salary

    def display_info(self):
        total_salary = self.calculate_salary()
        print(f"Name: {self.name}")
        print(f"Base Salary: {self.base_salary}")
        print(f"Experience: {self.experience} years")
        print(f"Skills: {self.skills}")
        print(f"Final Salary: {total_salary}")
        

# Object
s1 = Student("Nikhil", 30000, 3, ["Python", "Java"])
s1.display_info()
s1 = Student("Mohit", 40000, 4, ["Python", "Java"])
s1.display_info()
s1 = Student("Aditya", 50000, 3, ["Data Science", "Java"])
s1.display_info()
```

```python
# instance method

class Employee:
    comapany="Techno"

    def __init__(self,name,salary):
        self.name=name
        self.salary=salary

    def show_details(self):
        print(f"{self.name} works at {Employee.comapany} with salary {self.salary}")

    def apply_increment(self,percent):
        self.salary+=self.salary*(percent/100)
        print(f"New Salary for {self.name} is {self.salary}")

emp1=Employee("Nikhil",6000)
emp1.show_details()
emp1.apply_increment(10)
```

```python
# class Method

class Employee:
    company="Amazon"

    def __init__(self,name,salary):
        self.name=name
        self.salary=salary

    @classmethod
    def change_company(cls,new_company):
        cls.company=new_company

    def show_details(self):
        print(f"{self.name} works at {Employee.company} with salary {self.salary}")

emp1=Employee("Nikhil",90000)
emp1.show_details()

Employee.change_company("Google")
emp1.show_details()
```

```python
# Combined Example-All Three Methods Together

class Employee:
    company="TechCorp"
    conversion_rate=83.25

    def __init__(self,name,salary_used):
        self.name=name
        self.salary_used=salary_used

    # Instance mathod
    def show_salary_in_inr(self):
        inr=Employee.convert_currency(self.salary_used,Employee.conversion_rate)
        print(f"{self.name}'s salary in INR:{inr}")

    @classmethod
    def update_conversion_rate(cls,new_rate):
        cls.conversion_rate=new_rate
        print(f"Update conversion rate:{cls.conversion_rate}")

    @staticmethod
    def convert_currency(amount,rate):
        return amount*rate

emp1=Employee("Ram",3000)
emp2=Employee("Lakshman",4000)

emp1.show_salary_in_inr()
Employee.update_conversion_rate(85)
emp2.show_salary_in_inr()

converted=Employee.convert_currency(500,Employee.conversion_rate)
print("Direct conversion of $500 to INR:",converted)
```

```python
class Employee:
    def __init__(self,name,salary):
        self.__name=name
        self.__salary=salary

    # Getter
    def get_salary(self):
        return self.__salary
    
    # Setter
    def set_salary(self,new_salary):
        if new_salary<30000:
            print("Salary too low.Not allowed")
        else:
            self.__salary=new_salary
            print("Salary updated successfully!")
emp=Employee("Nikhil",49000)
print("Initial Salary:",emp.get_salary())

emp.set_salary(25000)
emp.set_salary(90000)
print("Updated Salary:",emp.get_salary())
```

```python
class Student:
    def __init__(self, name, age):
        self._name = name       # protected
        self._age = age         # protected

    # Getter for name
    def get_name(self):
        return self._name

    # Setter for name
    def set_name(self, new_name):
        self._name = new_name

    # Getter for age
    def get_age(self):
        return self._age

    # Setter for age (with validation)
    def set_age(self, new_age):
        if new_age > 0:
            self._age = new_age
        else:
            print("Invalid age")

# Using the class
s1 = Student("Nikhil", 22)

print(s1.get_name())   # Getter
s1.set_name("Rahul")   # Setter

print(s1.get_age())    # Getter
s1.set_age(25)         # Setter (valid)
s1.set_age(-10)        # Invalid
```

```python
# Bank Account

class BankAccount:
    def __init__(self,initial_balance,pin):
        self.__balance=initial_balance
        self.__pin=pin


# Deposite Money
    def deposite(self,amount):
        if amount>0:
            self.__balance+=amount
            print("Deposite Successfully")
        else:
            print("Invalid Deposite Amount")

# Withdraw Money
    def withdraw(self,amount,pin):
        if pin!=self.__pin:
            print("Invalid pin")
            return
        
        if amount > self.__balance:
            print("Insufficient funds")
        else:
            self.__balance-=amount
            print("Withdrwal successfully!")

# check Balance
    def check_balance(self,pin):
        if pin==self.__pin:
            return self.__balance
        else:
            print("Invalid PIN")
            return None
    
    def add_interest(self,rate):
        self.__balance+=self.__balance*rate
        print("Interested added.")


account=BankAccount(10000,224176)
account.deposite(-500)
account.deposite(500)
print("Balance:",account.check_balance(224176))
account.withdraw(2000,224176)
print("Balance:",account.check_balance(224176))
account.withdraw(20000,201316)
# account.withdraw(17000,224176)
account.add_interest(0.06)
print("Balance:",account.check_balance(224176))
```
