```python
class Car:
    def __init__(self,color,model):
        self.model=model
        self.color=color
        self.speed=0

    def start(self):
        print(f"{self.model} started.")

    def accelerate(self):
        self.speed+=10
        print(f"{self.speed} is now at {self.speed} km/h")


my_car=Car("Red","Tasala Model 3")

print(my_car.color)
my_car.start()
my_car.accelerate()
```
