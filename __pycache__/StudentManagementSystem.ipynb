```python
name=["Nikhil","Mohit","Aditya","Devender"]
age=[21,22,23,24]
grade=["A","A+","A","A-"]

def show_student(index):
    print(f"Name:{name[index]},Age:{age[index]},Grade:{grade[index]}")

print("Procedural style")
show_student(0)
show_student(1)
show_student(2)

print("==="*30)

class Student:
    name=""
    age=0
    grade=""

print("OOP style")
s1=Student()
s1.name="Nikhil"
s1.age=21
s1.grade="A"

s2=Student()
s2.name="Mohit"
s2.age=22
s2.grade="A+"

s3=Student()
s3.name="Aditya"
s3.age=23
s3.grade="A"

s4=Student()
s4.name="Devender"
s4.age=24
s4.grade="A-"

def show_student(student):
    print(f"Name:{student.name},Age:{student.age},Grade:{student.grade}")

show_student(s1)
show_student(s2)
show_student(s3)

```
