```python
# data is either numerical or categorical can be either discreate and continous
# Discrete data through its numerical but should be treated as categorical to the analysis
# but for model building all the categorical data should be converted to number
```

```python
cat_features=x.select_dtypes(include='object').columns
num_features=x.select_dtypes(exclude='object').columns
```

```python
cat_features
```

```python
# scaling>> numerical features
# categorical>> Onehat encoding
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler # both are data transfer
from sklearn.compose import ColumnTransformer

# ColumnTransformer applies tranformation to columns of array or pandas dataframe
# This estimator allows different columns subset of the input to the transformed
# will be concereated to form a single feature space.
# This is useful for hetrogeneous or columns data to combination several
# feature
# extraction mechanism or transformation into a single transformer
```

```python
numeric_transformer=StandardScaler()
oh_tranformer=OneHotEncoder(drop='first')
preprocessor=ColumnTransformer([
    ('OneHotEncoder',oh_tranformer,cat_features),
    ('StandardScaler',numeric_transformer,num_features)
])
```

```python
preprocessor
```

```python
preprocessor.fit_transform(x_train)
```
