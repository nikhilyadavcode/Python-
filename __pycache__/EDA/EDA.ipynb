```python
# 1 MB=1024 Kb
```

```python
def size_process(item):
    if str(item)[-1]=='M':
        res=float(str(item).replace("M",""))
        res=res*1024
        return res
    elif str(item)[-1]=='K':
        res=float(str(item).replace("K",""))
        return res
    else:
        return res(np.NaN)
```

```python
size_process("19M")
```

```python
size_process("19K")
```

```python
df_copy["size"]=df_copy["size"].apply(size_process)
```
