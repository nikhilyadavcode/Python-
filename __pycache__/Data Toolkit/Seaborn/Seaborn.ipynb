```python
# numerical data plotting
# relplot has options of plotting both categorical and numerical data
# relplot>>show the relationship between two variables
# relplot>> relationship plot
df
```

```python
# how tip is varing wrt total_bill>>scatterplot/by default relplot makes scatter plot
sns.relplot(x="total_bill",y="tip",data=df) # relationship between two variable,bu=y default it gives scatter plot
# insight>>as total bill increases and tip increases
```

```python
df
```

```python
sns.relplot(x="total_bill",y="tip",data=df,hue="smoker") # hue gives the different color sets
# insights>>customer who smoked gave the highest tip
```

```python
sns.relplot(x="total_bill",y="tip",data=df,hue="size")
```

```python
sns.relplot(x="total_bill",y="tip",data=df,style="size") # style gives the different representation of different column
```

```python
sns.relplot(x="total_bill",y="tip",data=df,style="smoker",hue="time")
```

```python
# relplot for making a line plot generating a data
# line plot is understand the trend in data,a time column/sequence is generally is there.
```

```python
time=np.arange(500)
value=np.random.randn(500).cumsum()
pd.DataFrame({"time":time,"value":value})
```

```python
sns.relplot(x="time", y="value", kind="line", data=df,sort=True)
```

```python
tips=sns.load_dataset("tips")
tips
```

```python
# Two subplots based on a column
sns.relplot(x="total_bill",y="tip",hue="smoker",data=tips,col="time")
```

```python
sns.relplot(x="total_bill",y="tip",hue="smoker",data=tips,col="size")
```

```python
# categorical data plot
df
```

```python
tips.dtypes
```

```python
# plot between data and total bill,day,is a categorical data
sns.catplot(x="day",y="total_bill",data=tips)
```

```python
# insights>>many people arrives on weekends as compared to weekdays.
# in cat plot,few datapoints are overlapping>> to see it seprately use swarm plot
# for categorical column>>cat plot
sns.catplot(x="day",y="total_bill",data=tips,kind="swarm")
```

```python
# boxplot >> to see outliers
sns.catplot(x="day",y="total_bill",data=tips,kind="box")
```

```python
sns.catplot(x="day",y="total_bill",data=tips,kind="violin")
```

```python
# boxen plot is made from box plot
sns.catplot(x="day",y="total_bill",data=tips,kind="boxen")
```

```python
sns.barplot(x="day",y="total_bill",data=tips,color="red")
```

```python
sns.jointplot(x=tips.total_bill,y=tips.tip) # draw a plot of two variables with distribution>>bivariate
```

```python
# joints plot>> understand relationship and also distribution of two variables
sns.jointplot(x=tips.total_bill,y=tips.tip,kind="reg") # regression line to represent relationship
```

```python
sns.jointplot(x=tips.total_bill,y=tips.tip,kind="hex")
```

```python
sns.jointplot(x=tips.total_bill,y=tips.tip,kind="kde") # kernal destination estimator,contour>>meterological department
```

```python
sns.pairplot(tips) # make plots with respect to all numerical columns
```

```python
sns.pairplot(tips,hue="smoker",palette="coolwarm")
```

```python
sns.pairplot(tips,hue="smoker")
```

```python
# boxplot using sns directory(earlier catplot)
sns.boxplot(x="day",y="total_bill",data=tips,palette="rainbow")
```

```python
sns.boxplot(data=tips,palette="rainbow",orient="h")
```
